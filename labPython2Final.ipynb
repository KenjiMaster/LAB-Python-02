## Ejercicio 6 - Minimax

```python
# Algoritmo Minimax para juego NIM limitado a escojer 1,2, o 3 pidras
CANTIDAD = 5 # Cantidad de montones
pilas = [i for i in range(1,CANTIDAD+1)]
memo = {} # Agregue memorizacion para que pudera tener montones mas grandes

# -- Sucesores, una lista con las jugadas en cierta pila
def sucesores(pilas):
  return [[j for j in range(1,4) if i-j >= 0] for i in pilas]

#-- Transicion, se indica la pila y la cantidad
def transicion(pilas,pila,piedras):
  newPilas = pilas.copy()
  newPilas[pila] -= piedras
  return newPilas

#-- Ganador, se gana cuando no hay piedras
def ganador(pilas):
  return sum(pilas) == 0

#-- Utilidad, se gana utilidad cuando no hay piedras y el el turno del rival
def utilidad(pilas,jugador):
  if ganador(pilas) and jugador: return 1
  if ganador(pilas) and not jugador: return -1
  return 0

#-- Minimax, usa memorizacion, True es MAX y False MIN (MAX es la cpu)
def minimax(pilas,maximizar):
  if (tuple(pilas),maximizar) in memo:
    return memo[(tuple(pilas),maximizar)]
  if ganador(pilas):
    return utilidad(pilas,not maximizar)
  if maximizar:
    mejor = float('-inf')
    for pila,posibles in enumerate(sucesores(pilas)):
      for piedras in posibles:
        resultado = minimax(transicion(pilas,pila,piedras),False)
        mejor = max(mejor,resultado)
    memo[(tuple(pilas),maximizar)] = mejor
    return mejor
  else:
    mejor = float('inf')
    for pila,posibles in enumerate(sucesores(pilas)):
      for piedras in posibles:
        resultado = minimax(transicion(pilas,pila,piedras),True)
        mejor = min(mejor,resultado)
    memo[(tuple(pilas),maximizar)] = mejor
    return mejor

#-- Mejor movimiento
def mejor_movimiento(pilas):
  mejor,movimiento= float('-inf'),None
  for pila,posibles in enumerate(sucesores(pilas)):
      for piedras in posibles:
        resultado = minimax(transicion(pilas,pila,piedras),False)
        if resultado > mejor:
            mejor,movimiento = resultado,(pila,piedras)
  return movimiento

#-- Mostrar las pilas con guia de numeros
def mostrar_pilas(pilas):
  for i in range(CANTIDAD,0,-1):
    for j in pilas:
      if j >= i:
        print("O",end=" ")
      else:
        print(" ",end=" ")
    print()
  print("-"*(CANTIDAD*2-1))
  for i in range(CANTIDAD):
    print(f"{i}",end=" ")
  print()

# Ejecucion
print("Tablero Inicial")
mostrar_pilas(pilas)

while not ganador(pilas):
  print("\nJugada Humano")
  valido = False
  while not valido:
    try:
      pila = int(input("Ingresa el numero de pila: "))
      if not 0 <= pila <= CANTIDAD-1:
        raise ValueError
      piedras = int(input("Ingresa la cantidad de piedras 1,2 o 3: "))
      if not 1<=piedras<=3:
        raise ValueError
      if piedras > pilas[pila]: raise ValueError
      valido = True
    except(ValueError):
      print("Solo se aceptan numeros validos")

  pilas = transicion(pilas,pila,piedras)
  mostrar_pilas(pilas)
  if ganador(pilas):
    print("Ganaste")
    break

  print("\nJugada CPU")
  CPU = mejor_movimiento(pilas)
  pilas = transicion(pilas,CPU[0],CPU[1])
  mostrar_pilas(pilas)
  if ganador(pilas):
    print("Gano la CPU")
    break
```
