## Ejercicio 3 - Map

### Punto 1

```python
from math import pi
#a. Dada una lista de temperaturas en Celsius, convertirlas a Fahrenheit.
celsius = [1,3,5,7,9,11,13,15,14,16,18,20,2,4]
fahrenheit = list(map(lambda c : c*(9/5) + 32,celsius))

#b. Dadas dos listas con la misma cantidad de elementos, generar una nueva sumando cada elementos (primer elemento de lista 1 + primer elemento lista 2).
lista1 = [1,2,3,4,5]
lista2 = [6,7,8,9,10]
suma = list(map(lambda x,y : x+y,lista1,lista2))

#c. Dada una lista de radios de circulos, generar la lista de sus areas.
radios = [1,3,5,7,9,11,13,15,14,16,18,20,2,4]
areas = list(map(lambda x: (x**2)*pi,radios))

#d. Dada un diccionario, generar una lista de tuplas (nombre, valor+10%)
dic = {"pan": 1000, "leche": 2500, "café": 5000}
lista = list(map(lambda nombre,valor: (nombre,valor+valor*0.1),dic.keys(),dic.values()))

#e. Dada una matriz, multiplicar todos los elementos por 10.
matriz = [[1,2,3], [4,5,6], [7,8,9]]
factor = list(map(lambda fila : list(map(lambda x:x*10,fila)),matriz))
```

## Ejercicio 4 - Filter

### Punto 1

```python
#a. Dada una lista de palabras, filtrar solo aquellas con mas de 4 letras.
palabras = ["ave","pez","pantano","camion","marioneta","reconocer"]
masCuatro = list(filter(lambda p:len(p)>4,palabras))

#b. Dada una lista filtrar los elementos que son nulos (None).
lista = [1,2,None,24,36,None,None]
noNulos = list(filter(lambda e: e != None,lista))

#c. Dada una lista de palabras, filtrar aquellas que empiezan con una vocal.
empiezaVocal = list(filter(lambda p: p[0]in"aeiou",palabras))

#d. Dada una lista de palabras, filtrar aquellas que son palindromos.
palindromos = list(filter(lambda p:p==p[::-1],palabras))

#e. Dada una lista de numeros, filtrar los que terminan en 5.
numeros = [1,2,3,4,5,23,25,46,75,34,85]
terminanCinco = list(filter(lambda n:n%10 == 5,numeros))

#f. Dada una lista de diccionarios, filtrar los productos con precio mayor a 100.
productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150}
]
precioMayor = list(filter(lambda p:p["precio"]>100,productos))
```

## Ejercicio 4 - Reduce

### Punto 1

```python
from functools import reduce
#a. Dada una lista de numeros, calcula la multiplicación de estos.
numeros = [1,2,3,4,5,23,25,46,75,34,85]
a = reduce(lambda x,y:x*y,numeros,1)

#b. Dada una lista de palabras, concetenalas en una sola.
palabras = ["ave","pez","pantano","camion","marioneta","reconocer"]
b = reduce(lambda x,y:x+y,palabras)

#c. Dada una lista, hallar el mayor de todos, define una función auxiliar que dado dos numeros retorne el mayor
c = reduce(lambda x,y : x if x > y else y,numeros)
```
